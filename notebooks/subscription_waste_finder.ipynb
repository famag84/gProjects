# Budget Tracking: Finding Subscription Creep

Tracking monthly subscriptions manually. Too many $5-10 charges add up.
This aggregates them and finds which services I'm not using.

```python
monthly_subscriptions = {
    'Streaming Service A': {'cost': 14.99, 'last_used_days_ago': 3},
    'Cloud Storage B': {'cost': 9.99, 'last_used_days_ago': 45},
    'Game Pass C': {'cost': 12.99, 'last_used_days_ago': 120},
    'Music App D': {'cost': 10.99, 'last_used_days_ago': 7},
    'VPN Service E': {'cost': 4.99, 'last_used_days_ago': 90},
}
```

```python
total_cost = sum(sub['cost'] for sub in monthly_subscriptions.values())
print(f'Total monthly cost: ${total_cost:.2f}\n')

unused_threshold = 30
waste_list = []

for name, info in monthly_subscriptions.items():
    if info['last_used_days_ago'] > unused_threshold:
        waste_list.append((name, info['cost'], info['last_used_days_ago']))

print('Services unused 30+ days (wasting money):')
waste_total = 0
for name, cost, days in waste_list:
    print(f'  {name}: ${cost:.2f} (last used {days} days ago)')
    waste_total += cost

print(f'\nPotential savings: ${waste_total:.2f}/month = ${waste_total * 12:.2f}/year')
```
