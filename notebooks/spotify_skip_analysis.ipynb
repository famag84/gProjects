# Analyzing My Spotify Listening Patterns

I exported my Spotify data and wanted to see which songs I skip most often.
This finds songs I start but never finish - my personal 'skip list'.

```python
# Simulate some listening data
listening_events = [
    {'track': 'Song A', 'played_seconds': 180, 'total_seconds': 200},
    {'track': 'Song B', 'played_seconds': 45, 'total_seconds': 210},
    {'track': 'Song A', 'played_seconds': 195, 'total_seconds': 200},
    {'track': 'Song C', 'played_seconds': 30, 'total_seconds': 180},
    {'track': 'Song B', 'played_seconds': 50, 'total_seconds': 210},
]
```

```python
track_stats = {}

for event in listening_events:
    track_name = event['track']
    completion_rate = (event['played_seconds'] / event['total_seconds']) * 100
    
    if track_name not in track_stats:
        track_stats[track_name] = []
    track_stats[track_name].append(completion_rate)

print('Skip rate by song:')
for track, rates in track_stats.items():
    avg_completion = sum(rates) / len(rates)
    print(f'{track}: {avg_completion:.1f}% completion average')
```

## Finding Skip Patterns

Songs below 60% completion are ones I consistently skip.
This helps clean my playlists.
